## Завдання 1. Додавання, віднімання та множення векторів на число

Дано вектори:
```python
a = np.array([7, -2, 5, 9])
```

## Завдання 2. Скалярний добуток і довжина векторів

## Завдання 3. Основні дії з матрицями

Дано матриці:
```python
A = np.array([
```

## Завдання 5. Циклічний зсув вектора

Дано вектор:
```python
v = np.array([12, 7, 19, 4, 11, 8])
```

1. Виконайте циклічний зсув на 2 позиції вправо.
2. Виконайте циклічний зсув на 3 позиції вліво.
3. Відновіть початковий вектор із результату першого зсуву.

Використайте `np.roll()`.

````python
# Заданий вектор
v = np.array([12, 7, 19, 4, 11, 8])

# 1. Циклічний зсув на 2 позиції вправо (додатне значення зсуву)
shift_right_2 = np.roll(v, 2)
print("Зсув на 2 позиції вправо:\n", shift_right_2)

# 2. Циклічний зсув на 3 позиції вліво (від'ємне значення зсуву)
shift_left_3 = np.roll(v, -3)
print("Зсув на 3 позиції вліво:\n", shift_left_3)

# 3. Відновлення початкового вектора з результату першого зсуву
# Оскільки робили зсув вправо на 2, для відновлення робимо зсув вліво на 2 (-2)
restored_v = np.roll(shift_right_2, -2)
print("Відновлений початковий вектор:\n", restored_v)
````

## Завдання 11. Перестановка рядків і стовпців

Дано матрицю:
```python
A = np.array([
    [11, 12, 13, 14],
    [21, 22, 23, 24],
    [31, 32, 33, 34],
    [41, 42, 43, 44]
])

row_order = np.array([2, 0, 3, 1])
col_order = np.array([3, 1, 0, 2])
```

1. Переставте рядки відповідно до `row_order`.
2. Після цього переставте стовпці відповідно до `col_order`.
3. За допомогою обернених перестановок відновіть початкову матрицю.

Для побудови оберненої перестановки використайте `np.argsort()`.

````python
# Задана матриця та вектори порядку
A = np.array([
    [11, 12, 13, 14],
    [21, 22, 23, 24],
    [31, 32, 33, 34],
    [41, 42, 43, 44]
])

row_order = np.array([2, 0, 3, 1])
col_order = np.array([3, 1, 0, 2])

# 1. Перестановка рядків відповідно до row_order
A_permuted_rows = A[row_order, :]

# 2. Перестановка стовпців відповідно до col_order
A_permuted = A_permuted_rows[:, col_order]
print("Матриця після перестановки рядків і стовпців:\n", A_permuted)

# 3. Відновлення початкової матриці за допомогою обернених перестановок
inv_row_order = np.argsort(row_order)
inv_col_order = np.argsort(col_order)

A_restored = A_permuted[inv_row_order, :][:, inv_col_order]
print("\nВідновлена початкова матриця:\n", A_restored)
````

## Завдання 12. Трикутні частини матриці

Дано матрицю:
```python

    [4, 1, 7, 2],

```

2. Отримайте нижню трикутну частину разом із головною діагоналлю.
3. Отримайте строго верхню та строго нижню частини.
4. Відновіть початкову матрицю з нижньої та строго верхньої частин.

Використайте `np.triu()` та `np.tril()`.

````python
# Задана матриця
A = np.array([
    [4, 1, 7, 2],
    [6, 3, 5, 8],
    [9, 2, 6, 1],
    [5, 7, 4, 3]
])

# 1. Верхня трикутна частина разом із головною діагоналлю
upper_tri = np.triu(A, k=0)
print("Верхня трикутна частина (з діагоналлю):\n", upper_tri)

# 2. Нижня трикутна частина разом із головною діагоналлю
lower_tri = np.tril(A, k=0)
print("\nНижня трикутна частина (з діагоналлю):\n", lower_tri)

# 3. Строго верхня та строго нижня частини (виключаючи головну діагональ)
strict_upper = np.triu(A, k=1)
strict_lower = np.tril(A, k=-1)
print("\nСтрого верхня частина:\n", strict_upper)
print("\nСтрого нижня частина:\n", strict_lower)

# 4. Відновлення початкової матриці
A_restored = lower_tri + strict_upper
print("\nВідновлена початкова матриця:\n", A_restored)
````

## Завдання 14. Добуток Кронекера

Дано матриці:
```python
A = np.array([
    [2, 1],
    [0, 3]
])

B = np.array([
    [1, 4],
    [2, 0]
])
```

1. Обчисліть добуток Кронекера `A` і `B`.

````python
# Перша пара матриць
A1 = np.array([[3, 0],
               [0, 7]])
B1 = np.array([[2, 0],
               [0, 5]])

# Друга пара матриць
A2 = np.array([[0, 2],
               [-1, 0]])
B2 = np.array([[4, 0],
               [0, 1]])

# 1. Обчислення комутатора для першої пари: [A1, B1] = A1 @ B1 - B1 @ A1
commutator_1 = (A1 @ B1) - (B1 @ A1)
print("Комутатор першої пари:\n", commutator_1)

# Перевірка комутативності (AB = BA) для першої пари
is_commutative_1 = np.allclose(A1 @ B1, B1 @ A1)
print("Чи виконується A1 @ B1 = B1 @ A1 для першої пари?:", is_commutative_1)

# 2. Обчислення комутатора для другої пари: [A2, B2] = A2 @ B2 - B2 @ A2
commutator_2 = (A2 @ B2) - (B2 @ A2)
print("\nКомутатор другої пари:\n", commutator_2)

# Перевірка комутативності для другої пари
is_commutative_2 = np.allclose(A2 @ B2, B2 @ A2)
print("Чи виконується A2 @ B2 = B2 @ A2 для другої пари?:", is_commutative_2)
````

```python
A = np.array([
```

```python
B = np.array([
```

## Завдання 19. Власні значення та власні вектори

Дано симетричну матрицю:
```python
A = np.array([
    [6.0, 2.0],
    [2.0, 3.0]
])
```

1. Знайдіть власні значення та власні вектори.
2. Перевірте співвідношення `A @ V = V * λ`.
3. Виведіть результат перевірки через `np.allclose()`.

Використайте `np.linalg.eigh()`.

````python
# Задана симетрична матриця
A = np.array([
    [6.0, 2.0],
    [2.0, 3.0]
])

# 1. Знаходження власних значень та власних векторів для симетричної матриці
eigenvalues, eigenvectors = np.linalg.eigh(A)
print("Власні значення (λ):\n", eigenvalues)
print("\nВласні вектори (V):\n", eigenvectors)

# 2. Перевірка співвідношення A @ V = V * λ
# У NumPy поелементне множення або множення матриці на діагональну матрицю
# Лівa частина: A @ V
left_side = A @ eigenvectors

# Права частина: V * eigenvalues (множить кожен стовпчик V на відповідне власне значення)
right_side = eigenvectors * eigenvalues

# 3. Перевірка через np.allclose()
is_correct = np.allclose(left_side, right_side)
print("\nРезультат перевірки A @ V = V * λ (np.allclose):", is_correct)
````

## Завдання 20. Метод найменших квадратів

Для точок
```python
x = np.array([0., 1., 2., 3., 4.])
y = np.array([2.2, 3.8, 6.1, 7.9, 10.2])
```

побудуйте пряму
\[
y \approx a + bx.
\]

1. Сформуйте матрицю ознак зі стовпця одиниць і значень `x`.
2. Знайдіть коефіцієнти `a` і `b` методом найменших квадратів.
3. Обчисліть прогнозні значення `y`.
4. Знайдіть суму квадратів залишків.

Використайте `np.linalg.lstsq()`.

````python
# Задані точки
x = np.array([0., 1., 2., 3., 4.])
y = np.array([2.2, 3.8, 6.1, 7.9, 10.2])

# 1. Створення матриці ознак (Design matrix), що складається зі стовпця одиниць (для вільного члена a) та x
X_mat = np.vstack([np.ones_like(x), x]).T
print("Матриця ознак X:\n", X_mat)

# 2. Знаходження коефіцієнтів a і b методом найменших квадратів
# np.linalg.lstsq повертає масив коефіцієнтів [a, b], залишки, ранг та сингулярні значення
coeffs, residuals, rank, s = np.linalg.lstsq(X_mat, y, rcond=None)
a, b = coeffs
print(f"Знайдені коефіцієнти: a = {a:.4f}, b = {b:.4f}")

# 3. Обчислення прогнозних значень y_pred
y_pred = X_mat @ coeffs
print("Прогнозні значення y:", y_pred)

# 4. Обчислення суми квадратів залишків
# Якщо residuals порожній (для надлишкових систем він повертається автоматично), обчислюємо вручну
sum_residuals = residuals[0] if residuals.size > 0 else np.sum((y - y_pred) ** 2)
print("Сума квадратів залишків:", sum_residuals)
````
